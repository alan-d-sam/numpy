- create a numpy array containing the marks of 10 students (3 subjects). print its shape,size,number of dimensions and datatype

```python
import numpy as np

marks = np.array([
    [85, 78, 92],
    [76, 88, 81],
    [90, 95, 89],
    [67, 72, 70],
    [82, 79, 85],
    [91, 87, 94],
    [73, 68, 75],
    [88, 91, 86],
    [79, 84, 80],
    [95, 93, 97]
])
print(marks)
```

Given the matrix:


```python
mat = np.arange(10,260,10).reshape(5,5)
print(mat)
```

- extract the first 3 rows and last 2 columns

```python
mat_row_col = mat[:3, 3:]
print(mat_row_col)
```

- extract middle matrix 3x3

```python
mat_ex_mid = mat[1:4, 1:4]
print(mat_ex_mid)
```

- extract all rows and alternate columns

```python
mat_ex_alt = mat[:,::2]
print(mat_ex_alt)
```

- extract last three rows and reverse the column order

```python
mat_last_3 = mat[2:, ::-1]
print(mat_last_3)
```

- extract the first two roes and reverse their elements

```python
mat_first_2 = mat[:2, ::-1]
print(mat_first_2)
```

Given matrix of 5 students :

```python
stud = np.array([
    [78,85,90],
    [45,60,55],
    [92,88,95],
    [35,40,30],
    [65,70,75]
])
bonus = np.array([5,3,2])
new_stud = np.array([[80,90,85]])
print(stud)
print(bonus)
print(new_stud)
```

```python
updated_mark = stud + bonus
print(updated_mark)
```

```python
ex_greater_80 = updated_mark[updated_mark > 80]
print(ex_greater_80)
```

```python
stud_less_40 = updated_mark[np.any(updated_mark < 40,axis=1)]
print(stud_less_40)
```

```python
avg = stud.mean(axis=1)
print(avg)
```

```python
final_mark = np.append(updated_mark, new_stud, axis=0)
print(final_mark)
```

```python
highest_mark = np.max(final_mark,axis=0)
print(highest_mark)
```

```python
tot_mark = np.sum(final_mark,axis=1)
print(tot_mark)
```

```python
desc_tot_mark = np.sort(tot_mark)[::-1]
print(desc_tot_mark)
```
