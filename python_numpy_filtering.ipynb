```python
a, b
```

```python
a + b
```

### broadcasting is a powerful mechanism in NumPy that allows arithmetic operations to be performed on arrays of different shapes.

###Normally, element-wise operations require both arrays to have the exact same dimensions. With broadcasting, NumPy virtually expands or stretches the smaller array so it matches the shape of the larger array, allowing the calculation to succeed without manually duplicating any data.

```python
a = np.arange(10, 28).reshape(-1, 3)
```

```python
a.size
```

```python
a
```

```python
np.max(a, axis=0)     #axis=0 means column wise
```

```python
np.sum(a, axis=1)     #axis=1 means row wise
```

```python
np.mean(a, axis=1)
```

```python
np.argmax(a)
```

```python
np.argmax(a, axis=0)
```

```python
np.sort(a)
```
