```python
import numpy as np
```

```python
marks = [80,87,78,67,81]
print(f"Average : {np.mean(marks)}")
print(f"Maximun : {np.max(marks)}")
print(f"Minimum : {np.min(marks)}")
```

```python
max = marks[0]
min = marks[0]
for i in range(1,len(marks)):
  if max < marks[i]:
    max = marks[i]
  if min > marks[i]:
    min = marks[i]
print(f"Maximum : {max}")
print(f"Minimum : {min}")
```
